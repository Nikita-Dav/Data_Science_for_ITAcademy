# 18. Зависит ли выигрыш от количества посетителей матча? (https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.cov.html)

Для определения наличия зависимости между исходом и количеством посетителей вычисляем ковариацию (укажет направление) и коэффициент корреляции Пирсона (укажет направление и степень линейной зависимости) между посещаемостью и победой домашней команды.

```python

cov = table['attendance'].cov(table['home_team_win'])
cor = table['attendance'].corr(table['home_team_win'])
print(f'Ковариация между посещаемостью и победой домашней команды: {cov:.2f}')
print(f'Коэффициент корреляции между посещаемостью и победой домашней команды: {cor:.3f}')
```

Вывод: Коэффициент корреляции положительный, но близок к нулю. Это указывает на крайне слабую положительную линейную связь между посещаемостью и победой домашней команды.

# 19. Правда ли что большинство проигрышных домашних матчей приходятся на Субботу и Воскресенье?

Проверяем значения в столбце дней недели. Считаем общее количество поражений.
Далее оставляем матчи, сыгранные в выходные и считаем количество поражений в них. Сравниваем полученные значения.

```python
table['day_of_week'].unique()
```

```python
total_loss = table[table['home_team_loss'] == 1].shape[0]
weekend = table[table['day_of_week'].isin(['Saturday', 'Sunday'])]
weekend_loss = weekend[weekend['home_team_loss'] == 1].shape[0]
print(f'Доля поражений в выходные: {(weekend_loss / total_loss * 100):.1f}%')
```

Вывод: большинство поражений не приходятся на субботу и воскресенье.
